## 4. Find the months that are missing

`s.asfreq("MS")` puts a date-indexed series on one row per month ("MS" = month start). A month the series skipped shows up as a missing value, so `.isna().sum()` counts the gaps.

**Watch:** run this and read it.

```python
dates = pd.to_datetime(["2024-01-01", "2024-02-01", "2024-04-01", "2024-05-01"])
toy = pd.Series([3.1, 3.0, 2.9, 3.2], index=dates)     # March is missing

on_grid = toy.asfreq("MS")
print(on_grid)
print("missing months:", on_grid.isna().sum())          # 1
```
